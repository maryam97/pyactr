```python
args = Namespace(data_path=data_folder, chains=chains, tunes=tunes, draws=draws)
```

```python
def investigate_trace(trace_path, args):
    pm_model = create_model(args)
    trace = az.from_netcdf(trace_path)
    with pm_model:
        az.plot_trace(trace, var_names=['decay', 'noise', 'threshold', 'lf', 'le', 'mu_rt', 'mu_prob'])
        az.plot_posterior(trace, var_names=['decay', 'noise', 'threshold', 'lf', 'le', 'mu_rt', 'mu_prob'])
        print(az.summary(trace, hdi_prob=.95))
```

```python
investigate_trace(f'{param_folder}/trace_draws={draws}_tune={tunes}_chains={chains}.nc', args)
```
